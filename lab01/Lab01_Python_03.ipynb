# Primera parte

```python
import requests

# Llamado a una API pública
url = "https://jsonplaceholder.typicode.com/users"
response = requests.get(url)
usuarios = response.json()
# Procesamiento imperativo
nombres = []
for u in usuarios:
    nombres.append(u["name"])

print("Usuarios encontrados:")
for n in nombres:
    print("-", n)

print('\n 📍Solutions---')
print("Users with the email end .biz are: ")
users_biz = []
for i in usuarios:
  if i["email"].endswith("biz"):
    users_biz.append(i["name"])

for i in users_biz:
    print("-", i)
```

```python

import requests

def pokemon(nombre_pokemon):
  # Llamado a la API
  url = f"https://pokeapi.co/api/v2/pokemon/{nombre_pokemon}"
  respuesta = requests.get(url)

  # Verificación del estado de la respuesta
  if respuesta.status_code == 200:
      datos = respuesta.json()


      info = {
          "nombre": datos["name"],
          "altura": datos["height"],
          "peso": datos["weight"],
          "tipos": [t["type"]["name"] for t in datos["types"]],
          "habilidades": [h["ability"]["name"] for h in datos["abilities"]],
          "experiencia": datos["base_experience"]
      }

      # Mostrar información
      print("\n📘 Información del Pokémon")
      print(f"Nombre: {info['nombre'].capitalize()}")
      print(f"Altura: {info['altura']}")
      print(f"Peso: {info['peso']}")
      print(f"Tipos: {', '.join(info['tipos'])}")
      print(f"Habilidades: {', '.join(info['habilidades'])}")

      print('\n📍 Solutions ---')

      # Tipo
      if len(info['tipos']) > 1:
          print("Es un Pokémon de tipo mixto")
      else:
          print(f"Es un Pokémon de tipo: {info['tipos'][0]}")

      # Peso
      if info['peso'] >= 100:
          print("Es un Pokémon pesado")
      else:
          print("Es un Pokémon ligero")

      # Habilidad principal
      print(f"Su habilidad principal es {info['habilidades'][0]}")

      # Experiencia
      print(f"Base experience: {info['experiencia']}")

      return info
```
